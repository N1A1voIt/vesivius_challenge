## Metrics Testing

Test the evaluation metrics.

```python
from src.utils.helpers import dice_coefficient, fbeta_score

# Create dummy predictions and targets
pred = np.random.rand(256, 256) > 0.5
target = np.random.rand(256, 256) > 0.5

dice = dice_coefficient(pred.astype(float), target.astype(float))
f05 = fbeta_score(pred.astype(float), target.astype(float), beta=0.5)

print(f'Dice coefficient: {dice:.4f}')
print(f'F0.5 score: {f05:.4f}')
```

## Training Loop Test

Test a single training iteration.

```python
from src.utils.helpers import AverageMeter, EarlyStopping

# Test AverageMeter
loss_meter = AverageMeter()
for i in range(10):
    loss_meter.update(np.random.rand())

print(f'Average loss: {loss_meter.avg:.4f}')

# Test EarlyStopping
early_stop = EarlyStopping(patience=3, mode='min')
losses = [0.5, 0.4, 0.3, 0.35, 0.36, 0.37, 0.38]

for epoch, loss in enumerate(losses):
    should_stop = early_stop(loss)
    print(f'Epoch {epoch}: loss={loss:.2f}, counter={early_stop.counter}, stop={should_stop}')
```
